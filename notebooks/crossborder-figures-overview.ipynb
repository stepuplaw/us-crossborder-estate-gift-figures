# US cross-border estate, gift and reporting figures, 2015 to 2026

This notebook loads `data/figures.csv` and `data/section7520_rates.csv` and reproduces the tables in the README. Each row of `figures.csv` is one figure for one year, with the Revenue Procedure, statute or Federal Register document the amount came from.

```python
import pandas as pd

fig = pd.read_csv('../data/figures.csv')
rates = pd.read_csv('../data/section7520_rates.csv')
len(fig), len(rates)
```

## Figures down, years across

The nonresident alien estate credit is one row with no year, because IRC 2102(b)(1) fixes it at $13,000.

```python
pivot = fig.dropna(subset=['year']).astype({'year': int}).pivot(index='figure_id', columns='year', values='amount')
pivot
```

## Growth from 2015 to 2026

The FBAR maximums grew faster than the IRS-indexed amounts because of the one-time 2016 catch-up adjustment.

```python
growth = (pivot[2026] / pivot[2015] - 1).mul(100).round(1).sort_values(ascending=False)
growth.rename('percent change 2015 to 2026')
```

## The 2018 double Revenue Procedure

Rev. Proc. 2018-18 superseded three 2018 amounts from Rev. Proc. 2017-58 after the Tax Cuts and Jobs Act.

```python
fig[(fig['year'] == 2018) & fig['authority'].str.contains('2018-18')][['figure_id', 'amount', 'note']]
```
